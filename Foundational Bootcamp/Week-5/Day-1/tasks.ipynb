```python
import numpy as np
n = np.arange(1, 21)        
arr = n**2 + 1                
matrix = arr.reshape(4, 5)   
print(matrix)
```

```python
double_matrix = matrix * 2
print(double_matrix)
```

```python
count_replaced = np.sum(matrix % 5 == 0)
print("Numbers replaced:", count_replaced)
```

```python
import numpy as np
A = np.arange(1, 101).reshape(10, 10)
prime_rows = A[[1, 2, 4, 6], :]
print(prime_rows)
```

```python
B = A.copy()
B[:, ::2] = B[::-1, ::2]   # reverse columns at even indices (0,2,4,…)
print(B)
```

```python
C = A.copy()
np.fill_diagonal(C, 0)
print(C)
```

```python
border_sum = (
    np.sum(A[0, :]) + np.sum(A[-1, :]) +   # top + bottom rows
    np.sum(A[1:-1, 0]) + np.sum(A[1:-1, -1])  # left + right columns (excluding corners)
)
print("Border sum:", border_sum)
```

```python
rotated = np.rot90(A, -1)   # -1 → clockwise
print(rotated)
```

```python
arr = np.random.randint(1, 501, size=1000)
perfect_squares = np.sum(np.sqrt(arr).astype(int)**2 == arr)
print("Perfect squares:", perfect_squares)
```

```python
from sympy import isprime

prime_count = np.sum([isprime(x) for x in arr])
print("Prime numbers:", prime_count)
```

```python
arr_mod = arr.copy()
arr_mod[arr_mod % 7 == 0] = np.sqrt(arr_mod[arr_mod % 7 == 0])
```

```python
sorted_arr = np.sort(arr)
largest_gap = np.max(np.diff(sorted_arr))
print("Largest gap:", largest_gap)
```

```python
cumsum = np.cumsum(arr)
index_exceed = np.argmax(cumsum > 100000)   # first index where condition is True
print("First index exceeding 100000:", index_exceed)
```

```python
import numpy as np

# Random patient counts between 50–250
data = np.random.randint(50, 251, size=(30, 5))
print(data.shape)   # (30,5)
```

```python
mean_dept = np.mean(data, axis=0)
print("Department-wise Mean:", mean_dept)
```

```python
median_dept = np.median(data, axis=0)
print("Department-wise Median:", median_dept)
```

```python
std_dept = np.std(data, axis=0)
print("Department-wise Std Dev:", std_dept)
```

```python
highest_day = np.max(data, axis=0)
print("Highest Patient Day:", highest_day)
```

```python
lowest_day = np.min(data, axis=0)
print("Lowest Patient Day:", lowest_day)
```

```python
# Calculate thresholds
upper = mean_dept + 2*std_dept
lower = mean_dept - 2*std_dept

# Boolean mask for outliers
outliers = (data < lower) | (data > upper)
print("Outliers detected:\n", outliers)
```

```python
# Copy data to avoid overwriting
cleaned_data = data.copy()

for dept in range(data.shape[1]):
    cleaned_data[outliers[:, dept], dept] = mean_dept[dept]

print("Cleaned Data:\n", cleaned_data)
```
